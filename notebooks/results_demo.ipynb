```python
import os
import json
import numpy as np
import pandas as pd
from PIL import Image
import matplotlib.pyplot as plt
import scipy.stats as stats
```

```python
styles_metadata = pd.read_csv("../data/raw/styles.csv", delimiter=';').set_index("id")
images_metadata = pd.read_csv("../data/raw/images.csv", delimiter=';')
image_ids = np.sort(styles_metadata.index)
image_address = ["../data/raw/images/"+str(id)+".jpg" for id in image_ids]
image_address_dictionary = {int(k):v for k,v in zip(image_ids,image_address)}
```

```python
def parse_responses(responses_files, path):
    responses = {}
    for file_name in responses_files:
        if file_name != '.DS_Store':
            with open(f'{path}{file_name}', 'r') as file:
                tmp_response = json.load(file)
                generated_id = tmp_response['generated_id']
                responses[generated_id] = tmp_response 
    return responses
```

```python
def plot_results(responses):
    for generated_id, matches in responses.items():
        print("Image query:", generated_id, matches)
        
        address_query = str("../" + matches["query_image"])
        image_query = Image.open(address_query)
        fig, axes = plt.subplots(1, 5, figsize=(15, 5))  
        
        axes[0].imshow(image_query)
        axes[0].axis("off")
        axes[0].set_title("Query Image")
        
        for i, match in enumerate(matches['matches']):
            id = int(match['id'])
            image_response = Image.open(image_address_dictionary[id])
            
            axes[i + 1].imshow(image_response)
            axes[i + 1].axis("off")
            axes[i + 1].set_title(f"Match {i+1}: {round(match['score']*100, 2)}%")
        
        plt.tight_layout()
        plt.show()
```

```python
folders = ["female", "male", "girls", "boys", "unisex"]

for folder in folders:
    responses_files = os.listdir(f"../data/query/responses/{folder}")
    path = f"../data/query/responses/{folder}/"
    try:
        responses = parse_responses(responses_files, path)
        plot_results(responses)
    except:
        print(responses_files, path)
```

```python
local_runtime = {
    50:70.7,
    60:81.1,
    70:94,
    80:109.5,
    90:123.2,
    100: 131.0,
    110: 145.0,
    120: 157.0,
    130: 173.0,
    140: 183.0,
    150: 196.0,
    160: 208.0,
    170: 220.0,
    180: 232.0,
    190: 244.0,
    200: 259.0,
    210: 271.0,
    220: 283.0,
    230: 294.0
}
```

```python
runtime = pd.DataFrame(
    list(local_runtime.items()), 
    columns=["Quantity", "Runtime total"]
)
```

```python
runtime["Runtime per image"] = (runtime["Runtime total"]/runtime["Quantity"]).round(3)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Categories to plot
categories = ["Runtime total", "Runtime per image"]

# Loop through categories and plot in respective subplot
for i, category in enumerate(categories):
    ax = axes[i]  # Select subplot
    
    # Area chart
    ax.fill_between(runtime["Quantity"], runtime[category], color="blue", alpha=0.6)

    # Customization
    ax.set_facecolor("lightgray")  
    ax.set_xlabel("Quantity of images")
    ax.set_ylabel("Runtime (seconds)")
    ax.set_title(f"{category} vs Quantity")
    ax.grid(True, linestyle="--", alpha=0.5)

# Adjust layout
plt.tight_layout()

# Show plot
plt.show()
```

Linear regression to assume potential runtime if the model run serially:

```python
x = runtime["Runtime total"]
y = runtime["Quantity"]
slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)
assumed_quantity = np.arange(300, 3000, 100)
y_pred = slope * assumed_quantity + intercept
runtime_linear_regression = pd.DataFrame(
    [assumed_quantity, y_pred],
    index = ["Quantity", "Runtime total, serial"]
).T
```

Amdahl's Law:

```python
def estimate_parallel_time(T_local, P_cluster, f):
    speedup = 1 / ((1 - f) + (f / P_cluster))
    T_parallel = T_local / speedup
    return T_parallel

T_local = 70.7  # Local execution time in seconds
P_cluster = 80  # Total cores in Databricks cluster
f = 0.9  # Fraction of workload that is parallelizable

T_parallel = estimate_parallel_time(T_local, P_cluster, f)
```

```python
runtime_linear_regression["Runtime total, parallel"] = runtime_linear_regression["Runtime total, serial"].apply(
    lambda x: estimate_parallel_time(x, 80, 0.9)
)
```

```python
runtime_linear_regression.columns
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.fill_between(
    runtime_linear_regression["Quantity"], 
    runtime_linear_regression["Runtime total, serial"], 
    color="blue", alpha=0.6, label="Runtime total, serial"
)
ax.fill_between(
    runtime_linear_regression["Quantity"], 
    runtime_linear_regression["Runtime total, parallel"], 
    color="red", alpha=0.6, label="Runtime total, parallel"
)

ax.set_facecolor("lightgray")  
ax.set_xlabel("Quantity of images")
ax.set_ylabel("Runtime (seconds)")
ax.set_title("Comparison of Serial vs Parallel Runtime")
ax.grid(True, linestyle="--", alpha=0.5)
ax.legend()

plt.show()
```
